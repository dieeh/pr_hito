```python
import sqlite3
import pandas as pd

DB_PATH = "inventario-2.db"

conn = sqlite3.connect(DB_PATH)

# Obtener todas las tablas
tablas = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%';",
    conn
)

print("\n=== TABLAS EN LA BD ===")
print(tablas)

# Mostrar contenido de cada tabla
for t in tablas["name"]:
    print(f"\n\n===== {t.upper()} =====")
    df = pd.read_sql_query(f"SELECT * FROM {t} LIMIT 50;", conn)
    print(df)

conn.close()
```
